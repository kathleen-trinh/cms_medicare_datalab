# README
This notebook tests your connection to your PostgreSQL database.
When prompted, please input the following:
- Database name
- Username
- Password
- Host name
- Post number

If the connection is successful, the database's version will be printed to the console.

```python
import getpass
import psycopg2
import pandas as pd
from IPython.display import display, HTML
```

```python
DATABASE = input("Database: ")
```

```python
USERNAME = input("Username: ")
```

```python
PASSWORD = getpass.getpass("Password: ")
#print("Password entered (hidden):", PASSWORD)
```

```python
HOST = input("Host: ")
```

```python
PORT_NUMBER = input("Port Number: ")
```

```python
try:
    connection = psycopg2.connect(database=DATABASE, user=USERNAME, password=PASSWORD, host=HOST, port=PORT_NUMBER)
    cursor = connection.cursor()
    cursor.execute("SELECT version();")
    record = cursor.fetchone()
    print("Database version:", record)
except psycopg2.Error as error:
    print("Error while connecting to PostgreSQL database", error)
finally:
    if cursor:
        cursor.close()
    if connection:
        connection.close()
        print("PostgreSQL connection is closed.")
```

```python
try:
    connection = psycopg2.connect(database=DATABASE, user=USERNAME, password=PASSWORD, host=HOST, port=PORT_NUMBER)
    query = "SELECT * FROM cms_raw.beneficiaries_2008 LIMIT 10;"
    df = pd.read_sql_query(query, connection)
    display(HTML(df.to_html()))
except psycopg2.Error as error:
    print("Error while connecting to PostgreSQL database", error)
finally:
    if connection:
        connection.close()
        print("PostgreSQL connection is closed.")
```
